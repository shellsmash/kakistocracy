```python
from transformers import pipeline
```

```python
pipe = pipeline("text-generation", model="google/gemma-3-270m-it", device_map="auto")

# 3. Start with a completely empty history list
messages = []
```

```python
# 4. Interactive chat loop
while True:
    user_input = input("You: ")

    if user_input.lower() in ['quit', 'exit']:
        print("AI: Goodbye!")
        break

    if not user_input.strip():
        continue

    # Turn 1, 3, 5... (Always 'user')
    messages.append({"role": "user", "content": user_input})

    # Send the alternating history to the model
    raw_output = pipe(messages, max_new_tokens=256)

    # print(raw_output)

    # Extract the clean response text
    ai_response = raw_output[0]["generated_text"][-1]["content"]

    print(f"AI: {ai_response}\n")

    # Turn 2, 4, 6... (Always 'assistant')
    messages.append({"role": "assistant", "content": ai_response})
```
